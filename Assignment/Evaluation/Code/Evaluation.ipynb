```python
#This code is to deal with version inconsistency
#Otherwise raises
#ValueError: mutable default <class 'esm.esmfold.v1.trunk.StructureModuleConfig'> for field structure_module is not allowed: use default_factory
#https://github.com/facebookresearch/esm/issues/696
#https://github.com/asarigun/esm/tree/main/esm/esmfold/v1
#Based on the comit messages, there should be debugging in esmfold.py&trunk.py
from pathlib import Path
import esm

esm_dir = Path(esm.__file__).resolve().parent
print("ESM package dir:", esm_dir)

esmfold_py = esm_dir / "esmfold" / "v1" / "esmfold.py"
txt = esmfold_py.read_text()
if "from dataclasses import dataclass, field" not in txt:
    txt = txt.replace(
        "from dataclasses import dataclass",
        "from dataclasses import dataclass, field")
txt = txt.replace(
    "trunk: T.Any = FoldingTrunkConfig()",
    "trunk: T.Any = field(default_factory=FoldingTrunkConfig)")
esmfold_py.write_text(txt)

trunk_py = esm_dir / "esmfold" / "v1" / "trunk.py"
txt = trunk_py.read_text()
if "from dataclasses import dataclass, field" not in txt:
    txt = txt.replace(
        "from dataclasses import dataclass",
        "from dataclasses import dataclass, field")
txt = txt.replace(
    "structure_module: StructureModuleConfig = StructureModuleConfig()",
    "structure_module: StructureModuleConfig = field(default_factory=StructureModuleConfig)")
trunk_py.write_text(txt)
print("Patched:", esmfold_py)
print("Patched:", trunk_py)
```

```python
df = pd.DataFrame(rows)
df.to_csv("/content/fast.csv", index=False)
files.download("/content/fast.csv")
```
